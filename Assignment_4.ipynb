```python
# Q1

import pandas as pd

fixed_entries=[
    {"question": "what is the annual fee",
     "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge",
     "category": "billing"},

    {"question": "how to reset password",
     "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login",
     "category": "account"},

    {"question": "what are your working hours",
     "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time",
     "category": "general"},

    {"question": "how can i pay the fee",
     "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee",
     "category": "billing"}
]

roll_number=input("Enter your roll number: ")

last_two=roll_number[-2:]

entries=fixed_entries.copy()

categories=["billing","account","general"]

for digit in last_two:
    digit=int(digit)

    category=categories[digit % 3]

    if category=="billing":
        question="how can I pay my fee"
        answer="You can pay your fee using UPI."
        keywords="pay fee payment"

    elif category=="account":
        question="how do I update my registered mobile number"
        answer="You can update your mobile number from account settings."
        keywords="update mobile number account"

    else:
        question="what are the college office hours"
        answer="The college office is open from 9 AM to 5 PM."
        keywords="college office hours timing"

    entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })

df=pd.DataFrame(entries)

print(df)
```

```python
# Q2

def search_questions(query, df):
    query_words=query.lower().split()

    results=[]

    for i in range(len(df)):
        keywords=df.loc[i,"keywords"].lower().split()

        score=0
        
        for word in query_words:
            if word in keywords:
                score=score + 1

        if score > 0:
            results.append([df.loc[i,"question"],score])

    results.sort(key=lambda x: x[1], reverse=True)
    return results
    
query=input("Enter your question: ")

results=search_questions(query, df)

for result in results:
    print("Question:", result[0])
    print("Score:", result[1])
```

```python
# Q3

def same_category(category_name, df):
    
    result=df[df["category"]==category_name]
    return result["question"]

category_name=input("Enter category: ")

questions=same_category(category_name,df)

print("Questions in this category:")

print(questions)
```

```python
# Q4

number=int(input("Enter entry number from 0 to 5: "))

new_keyword=input("Enter new keyword: ")

df.loc[number,"keywords"]=df.loc[number,"keywords"]+" "+new_keyword

roll_number=input("Enter your roll number: ")

file_name=roll_number+"_faq_data.csv"

df.to_csv(file_name,index=False)

print("Keyword added successfully.")
print("File saved as:", file_name)
```

```python
# Q5

category_count = df.groupby("category").size()
print("Number of FAQ entries in each category:")
print(category_count)
```

```python
# Q6

def search_questions(query, df):
    query_words=query.lower().split()
    scores =[]

    for i in range(len(df)):
        keywords = df.loc[i,"keywords"].lower().split()
        score=0
        
        for word in query_words:
            if word in keywords:
                score=score+1
        scores.append(score)

    highest_score=max(scores)

    print("Highest score:",highest_score)

    for i in range(len(df)):
        if scores[i]==highest_score:
            print("Question:",df.loc[i,"question"])
            print("Answer:",df.loc[i,"answer"])
            print("Score:",scores[i])

query = input("Enter your question: ")
search_questions(query, df)
```
