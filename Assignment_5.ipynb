```python
#Q1

import numpy as np

arr=np.array([1,2,3,4,5])
print(arr+2)
print(arr*3)
print(arr/2)
```

```python
#Q2
#(a)
arr=np.array([1,2,3,6,4,5])
print("Reverse arr:",np.flip(arr))

#(b)(i)
x=np.array([1,2,3,4,5,1,2,1,1,1])
values,counts=np.unique(x,return_counts=True)
index=np.argmax(counts)
value=values[index]
print("Most frequent value of x:",value)
print("Indices of x:",np.where(x==value)[0])

#(b)(ii)
y=np.array([1,1,1,2,3,4,2,4,3,3])
values_y,counts_y=np.unique(y,return_counts=True)
index_y=np.argmax(counts_y)
value_y=values_y[index_y]
print("Most frequent value of y:",value_y)
print("Indices of y:",np.where(y==value_y)[0])
```
