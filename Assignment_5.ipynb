```python
#Q1

import numpy as np

arr=np.array([1,2,3,4,5])
print(arr+2)
print(arr*3)
print(arr/2)
```

```python
#Q2
#(a)
arr=np.array([1,2,3,6,4,5])
print("Reverse arr:",np.flip(arr))

#(b)(i)
x=np.array([1,2,3,4,5,1,2,1,1,1])
values,counts=np.unique(x,return_counts=True)
index=np.argmax(counts)
value=values[index]
print("Most frequent value of x:",value)
print("Indices of x:",np.where(x==value)[0])

#(b)(ii)
y=np.array([1,1,1,2,3,4,2,4,3,3])
values_y,counts_y=np.unique(y,return_counts=True)
index_y=np.argmax(counts_y)
value_y=values_y[index_y]
print("Most frequent value of y:",value_y)
print("Indices of y:",np.where(y==value_y)[0])
```

```python
#Q3

arr=np.array([[10,20,30],[40,50,60],[70,80,90]])

print("arr[0,1]:",arr[0,1])
print("arr[2,0]:",arr[2,0])
```

```python
#Q4

vansh=np.linspace(10,100,25)

print("vansh:",vansh)
print("Dimensions:",vansh.ndim)
print("Shape:",vansh.shape)
print("Total elements:",vansh.size)
print("Data type:",vansh.dtype)
print("Bytes:",vansh.nbytes)

vansh=vansh.reshape(25,1)

print(vansh)
print(vansh.T)
```

```python
#Q5

ucs420_vansh=np.array([[10,20,30,40],[50,60,70,80],[90,15,20,35]])

print(ucs420_vansh)
print("Mean:",np.mean(ucs420_vansh))
print("Median:",np.median(ucs420_vansh))
print("Maximum:",np.max(ucs420_vansh))
print("Minimum:",np.min(ucs420_vansh))
print("Unique values:",np.unique(ucs420_vansh))

reshaped=ucs420_vansh.reshape(4,3)
print("Reshaped:",reshaped)

resized=np.resize(ucs420_vansh,(2,3))
print("Resized:",resized)
```
