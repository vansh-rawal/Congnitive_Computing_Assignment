```python
#Q1

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

M=float(input("Enter the value of M: "))
x=np.linspace(-10,10,100)
y1=M*x**2
y2=M*np.sin(x)

plt.plot(x,y1,label="M*x^2")
plt.plot(x,y2,label="M*sin(x)",linestyle="--")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Mathematical Functions")
plt.legend()
plt.grid()

plt.show()
```

```python
#Q2

subjects=["Maths","Physics","Chemistry","English","Computer"]
scores=[85,78,92,75,88]

df=pd.DataFrame({"Subject":subjects,"Score":scores})
print(df)

sns.barplot(x="Subject",y="Score",data=df,hue="Subject",legend=False)

for i in range(len(scores)):
    plt.text(i,scores[i]+1,str(scores[i]),ha="center")

plt.title("Subject Scores")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis="y")

plt.show()
```

```python
#Q3

np.random.seed(1024170325)

data=np.random.rand(50)
x=np.arange(1,51)

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot(x,np.cumsum(data))
plt.title("Cumulative Sum")
plt.xlabel("Value")
plt.ylabel("Cumulative Sum")
plt.grid()

plt.subplot(2,2,2)
plt.scatter(x,data)
plt.title("Random Data")
plt.xlabel("Value")
plt.ylabel("Random Value")
plt.grid()

plt.subplot(2,2,3)
plt.plot(x,data)
plt.title("Random Values")
plt.xlabel("Value")
plt.ylabel("Value")
plt.grid()

plt.subplot(2,2,4)
plt.hist(data)
plt.title("Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid()

plt.tight_layout()

plt.show()
```
